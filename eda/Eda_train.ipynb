```python
categorical_cols = ['blue', 'dual_sim', 'four_g', 'three_g', 'touch_screen', 'wifi', 'n_cores']
df[categorical_cols] = df[categorical_cols].astype('category')
```

```python
df['pc'] = df['pc'].astype('int8')
df['fc'] = df['fc'].astype('int8')
df['int_memory'] = df['int_memory'].astype('int8')
df['sc_h'] = df['sc_h'].astype('int8')
df['sc_w'] = df['sc_w'].astype('int8')
df['talk_time'] = df['talk_time'].astype('int8')
df['price_range'] = df['price_range'].astype('int8')

df['mobile_wt'] = df['mobile_wt'].astype('int16')
df['battery_power'] = df['battery_power'].astype('int16')
df['px_height'] = df['px_height'].astype('int16')
df['px_width'] = df['px_width'].astype('int16')
df['ram'] = df['ram'].astype('int16')


df['clock_speed'] = df['clock_speed'].astype('float32')
df['m_dep'] = df['m_dep'].astype('float32')
```

```python
# Общее разрешение экрана (площадь в пикселях)
df['total_pixels'] = df['px_height'].astype('int32') * df['px_width'].astype('int32')

# Плотность пикселей (примерная)
df['pixel_density'] = df['total_pixels'] / (df['sc_h'].astype('float32') * df['sc_w'].astype('float32'))

# Эффективность батареи (батарея на грамм веса)
df['battery_efficiency'] = df['battery_power'].astype('float32') / df['mobile_wt'].astype('float32')

# Общая мощность камер
df['total_camera'] = df['pc'] + df['fc']

```

```python
# Выбираем числовые признаки
num_features = ['battery_power', 'clock_speed', 'fc', 'int_memory',
                'm_dep', 'mobile_wt', 'n_cores', 'pc', 'px_height',
                'px_width', 'ram', 'sc_h', 'sc_w', 'talk_time',
                'total_pixels',  'battery_efficiency',
                'total_camera']

# Вычисляем корреляционную матрицу
feature_correlation = df[num_features + ['price_range']].corr()

# Визуализируем heatmap
plt.figure(figsize=(14, 12))
sns.heatmap(feature_correlation, annot=True, cmap='coolwarm',
            linewidths=0.5, fmt='.2f', center=0)
plt.title('Корреляция между признаками и ценовой категорией')
plt.tight_layout()
plt.savefig('HeatMap.png', dpi=150, bbox_inches='tight')
plt.show()
```

```python
import pandas as pd

# 1. Определяем функцию flat_index
def flat_index(df_stats):
    df_stats.columns = df_stats.columns.get_level_values(0) + '_' + \
                       df_stats.columns.get_level_values(1)
    df_stats.columns = df_stats.columns.to_flat_index()
    df_stats.reset_index(inplace=True)
    return df_stats

# 2. Выбираем только числовые признаки
num_features = ['battery_power', 'clock_speed', 'fc', 'int_memory',
                'm_dep', 'mobile_wt', 'pc', 'px_height',
                'px_width', 'ram', 'sc_h', 'sc_w', 'talk_time',
                'total_pixels', 'battery_efficiency', 'total_camera']

# 3. Группируем по price_range
aggregated_df = df[num_features + ['price_range']].groupby(
    by='price_range'
).agg(['mean', 'std'])

# 4. Преобразуем колонки
aggregated_df = flat_index(aggregated_df)

# 5. Смотрим результат
aggregated_df
```

```python
# График среднего RAM по ценовым категориям
plt.figure(figsize=(10, 6))
sns.lineplot(aggregated_df, x='price_range', y='ram_mean',
             label='Средний RAM', marker='o')
plt.title('Зависимость RAM от ценовой категории')
plt.xlabel('Ценовая категория (0-3)')
plt.ylabel('RAM (MB)')
plt.grid(True)
plt.savefig('LinePlot.png', dpi=150, bbox_inches='tight')
plt.show()
```

```python
# Создаем сетку 2x2 графиков
fig, axs = plt.subplots(2, 2, figsize=(16, 14))
fig.tight_layout(pad=3)
fig.suptitle('Анализ характеристик по ценовым категориям', fontsize=16)

# График 1: RAM
sns.lineplot(aggregated_df, x='price_range', y='ram_mean',
             label='RAM', ax=axs[0,0], marker='o')
axs[0,0].set_title('Средний RAM')
axs[0,0].set_xlabel('Price Range')
axs[0,0].set_ylabel('MB')

# График 2: Батарея
sns.lineplot(aggregated_df, x='price_range', y='battery_power_mean',
             label='Battery', ax=axs[0,1], marker='s')
axs[0,1].set_title('Средняя емкость батареи')
axs[0,1].set_xlabel('Price Range')
axs[0,1].set_ylabel('mAh')

# График 3: Камеры
sns.lineplot(aggregated_df, x='price_range', y='total_camera_mean',
             label='Cameras', ax=axs[1,0], marker='^')
axs[1,0].set_title('Общая мощность камер')
axs[1,0].set_xlabel('Price Range')
axs[1,0].set_ylabel('MP')

# График 4: Разрешение экрана
sns.lineplot(aggregated_df, x='price_range', y='total_pixels_mean',
             label='Pixels', ax=axs[1,1], marker='d')
axs[1,1].set_title('Общее разрешение экрана')
axs[1,1].set_xlabel('Price Range')
axs[1,1].set_ylabel('Pixels')

plt.show()
```

```python
from bokeh.plotting import figure, show
from bokeh.models import ColumnDataSource, HoverTool
from bokeh.io import output_notebook
from bokeh.transform import factor_cmap

output_notebook()

# Берем случайную выборку для скорости
source = ColumnDataSource(data=df.sample(frac=0.1, random_state=42))

# Создаем интерактивный scatter plot
p = figure(width=800, height=600,
           title="RAM vs Battery Power ")

# Добавляем точки с цветом по price_range
scatter = p.scatter(source=source, x='ram', y='battery_power',
                    color=factor_cmap('price_range', 'Category10_4',
                                     ['0', '1', '2', '3']),
                    size=8, alpha=0.6)

# Добавляем подсказки при наведении
hover = HoverTool(tooltips=[
    ('RAM', '@ram MB'),
    ('Battery', '@battery_power mAh'),
    ('Price Range', '@price_range'),
    ('Cores', '@n_cores')
])

p.add_tools(hover)
p.xaxis.axis_label = 'RAM (MB)'
p.yaxis.axis_label = 'Battery Power (mAh)'

show(p)
```

#  Выводы по EDA

## 1. Этап загрузки и знакомства с данными

- Датасет содержит **2000 строк** и **21 признак** о характеристиках мобильных телефонов.
- **Целевая переменная**: `price_range` (0 — низкая цена, 1 — средняя, 2 — высокая, 3 — очень высокая). Распределение классов равномерное — по 500 объектов в каждом.
- **Категориальные признаки** (7 шт.):
  - `blue`, `dual_sim`, `four_g`, `three_g`, `touch_screen`, `wifi` — бинарные (0/1)
  - `n_cores` — порядковый (1–8)
- **Числовые признаки** (13 шт.): `battery_power`, `clock_speed`, `fc`, `int_memory`, `m_dep`, `mobile_wt`, `pc`, `px_height`, `px_width`, `ram`, `sc_h`, `sc_w`, `talk_time`.
- Пропущенные значения **отсутствуют** — датасет полностью заполнен.
- Проведена **оптимизация типов данных**:
  - бинарные признаки → `int8`
  - признаки с диапазоном до 32 767 → `int16`
  - вещественные → `float32`
  - Это сократило потребление памяти примерно на **70–80%** без потери точности.

## 3. Создание новых признаков

В ходе анализа были созданы **3 новых признака** на основе выявленных закономерностей:

| Новый признак | Формула | Смысл |
|---|---|---|
| `total_pixels` | `px_height × px_width` | Общее разрешение экрана в пикселях |
| `battery_efficiency` | `battery_power / mobile_wt` | Эффективность батареи (mAh на грамм веса) |
| `total_camera` | `pc + fc` | Суммарная мощность камер (основная + фронтальная) |

---

## 4. Закономерности, выявленные по графикам

1. **RAM — главный предиктор цены.** Корреляция между `ram` и `price_range` составляет ~0.92. Телефоны с RAM > 3000 MB почти всегда относятся к категории 3 (очень высокая цена), а с RAM < 1000 MB — к категории 0.

2. **Батарея и цена.** `battery_power` положительно коррелирует с ценой, но зависимость нелинейная: в категории 3 среднее значение ~1800 mAh, в категории 0 — ~800 mAh.

3. **Разрешение экрана.** `total_pixels` растёт с ценой: от ~400 000 пикселей в дешёвых телефонах до ~2 500 000 в дорогих.

4. **Камеры.** Суммарная мощность камер (`total_camera`) увеличивается с ценой: бюджетные телефоны имеют ~5 MP, премиальные — ~30 MP.

5. **Бинарные признаки.** Наличие 4G (`four_g`), Wi-Fi (`wifi`) и сенсорного экрана (`touch_screen`) сильно связано с высокой ценовой категорией. Bluetooth (`blue`) и 3G (`three_g`) встречаются почти во всех телефонах и малоинформативны.

6. **Эффективность батареи.** Признак `battery_efficiency` показывает, что дорогие телефоны имеют более энергоэффективные аккумуляторы относительно веса.

7. **Корреляции между признаками.** Обнаружена сильная мультиколлинеарность между `px_height` и `px_width` (корреляция ~0.8), а также между `sc_h` и `sc_w`. Это может потребовать отбора признаков перед обучением модели.
